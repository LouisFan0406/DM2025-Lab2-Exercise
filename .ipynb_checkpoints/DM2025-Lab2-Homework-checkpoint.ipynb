### 1.1 Preprocessing Steps

Before training any model, I applied a structured preprocessing pipeline to clean and normalize the Twitter posts. Since the dataset contains short, informal, and very noisy text, preprocessing becomes important for helping TF-IDF extract stable linguistic patterns. Without basic cleaning, many meaningless tokens (e.g., URLs, user mentions, emojis, numbers) would dominate the vocabulary and weaken the model’s ability to learn emotion-related features.

The preprocessing steps include:

- **Lowercasing all text**  
  This reduces the vocabulary size by treating “Happy” and “happy” as the same word.

- **Removing URLs**  
  Many tweets contain links such as `http://`, `https://`, or `www.` which carry no emotional value. These are removed.

- **Removing usernames and hashtags**  
  Mentions like `@user123` or hashtags like `#moodtoday` often introduce noise or extremely specific tokens that do not generalize.

- **Removing all non-alphabetic characters**  
  Digits, emojis, punctuation, and mixed symbols are stripped to keep only letters.  
  This helps TF-IDF focus on textual patterns directly related to emotions.

- **Condensing multiple spaces**  
  After removal operations, spacing may become messy; collapsing whitespace ensures clean and consistent input.

Overall, this preprocessing pipeline reduces noisy patterns that interfere with TF-IDF and helps create a cleaner, more model-friendly dataset. Although simple, these steps significantly improve robustness and training stability.

### 1.2 Feature Engineering Steps

To convert raw text into numerical features, I experimented with different TF-IDF configurations. The two main representations I tested were word-level TF-IDF and character-level TF-IDF. After comparing several settings, the character-level representation consistently performed better.

### Final TF-IDF settings used:

- `analyzer = "char"`  
- Character n-grams ranging from **3 to 5**  
- Maximum features = **200,000**  
- Stopwords not removed (they do not affect char-level features)

Character-level TF-IDF works particularly well for short and informal social media posts. It captures internal character sequences like:

- `"ang"` (appears in *angry*, *anger*, *angrily*)  
- `"sad"`  
- `"lol"`  
- repeated letters like `"aaa"` or `"!!!"`

This subword-based representation is more resilient to:

- Misspellings  
- Slang  
- Abbreviations  
- Informal writing styles  
- Very short posts where full words may not carry meaning  

In contrast, word-level TF-IDF struggled with fragmented or noisy text, especially when posts had irregular vocabulary or consisted of only a few words. Character-level TF-IDF extracts more stable statistical patterns, which ultimately led to higher leaderboard scores.


### 1.3 Explanation of Your Model

After generating TF-IDF features, I tested multiple models and found that **LinearSVC** performed the best. The final model configuration was:

- `LinearSVC(C=1.0, class_weight="balanced")`

### Why this model works well:

- **Linear SVM handles high-dimensional sparse TF-IDF vectors efficiently.**  
  Since TF-IDF produces hundreds of thousands of features, LinearSVC is well-suited for this type of data.

- **Margin-based learning provides strong boundaries between emotion classes.**  
  Emotions often overlap (e.g., sadness vs. fear), so a model that maximizes separation helps improve macro F1.

- **Balanced class weights reduce bias toward frequent emotions.**  
  Some labels appear much more often than others. The balanced weighting helps stabilize predictions.

- **Combining SVM with char-level TF-IDF captures reliable emotional cues.**  
  Even when tweets are misspelled or contain partial words, the model still identifies meaningful character patterns.

Among all experiments—including Logistic Regression and word-level models—this configuration consistently achieved the highest performance on both local validation and Kaggle’s public leaderboard. Therefore, it was chosen as the final competition submission.

# **Competition Code**

## 1. Preprocessing Steps

This section performs all data loading and text preprocessing required for the competition.
I cleaned the raw tweet text by applying lowercasing, removing URLs, mentions, hashtags, punctuation, and reducing multiple spaces.
The goal of this preprocessing step is to normalize the inputs and reduce noise before feature extraction.

```python
### Add the code related to the preprocessing steps in cells inside this section
import re
import json
import pandas as pd

# Load final_posts.json
with open("final_posts.json", "r", encoding="utf-8") as f:
    data = json.load(f)

rows = []
for item in data:
    post = item["root"]["_source"]["post"]
    rows.append({
        "id": post["post_id"],
        "text": post["text"]
    })

posts = pd.DataFrame(rows)

# Load the other two CSV files
iden = pd.read_csv("data_identification.csv")   # id, split
emotion = pd.read_csv("emotion.csv")           # id, emotion

# Merge datasets
df = posts.merge(iden, on="id", how="left")
df = df.merge(emotion, on="id", how="left")

# Clean text function
def clean_text(s):
    s = s.lower()
    s = re.sub(r"http\S+|www\S+|https\S+", " ", s)
    s = re.sub(r"[@#]\w+", " ", s)
    s = re.sub(r"[^a-z\s]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

df["text"] = df["text"].astype(str).apply(clean_text)

# Split final train/test
train_df = df[df["split"] == "train"].copy()
test_df = df[df["split"] == "test"].copy()

print("Preprocessing complete.")
```

## 2. Feature Engineering Steps

I extracted two types of text features:

1.Word-level TF–IDF (1–2 grams)

2.Character-level TF–IDF (3–5 grams)

The motivation is to capture both semantic signals (word TF–IDF) and sub-word patterns, misspellings, and stylistic cues (character TF–IDF).
Char-TFIDF performed significantly better in validation and was the feature set used in my best submission.

```python
### Add the code related to the feature engineering steps in cells inside this section
from sklearn.feature_extraction.text import TfidfVectorizer

# Word TF-IDF
word_tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=200000,
    ngram_range=(1, 2)
)

# Character TF-IDF
char_tfidf = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3, 5),
    max_features=200000
)

print("Feature engineering objects created.")
```

## 3. Model Implementation Steps

Two models were trained:

1.Logistic Regression with word-level TF–IDF

2.Linear SVM with character-level TF–IDF

The Linear SVM model with char-TFIDF achieved the best performance and produced my final Kaggle submission.
I used class weighting, larger feature space, and a strong linear classifier to optimize macro-F1.

```python
### Add the code related to the model implementation steps in cells inside this section
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline

X_all = train_df["text"]
y_all = train_df["emotion"]
X_test = test_df["text"]

# ------- Model 1: Word TF-IDF + Logistic Regression -------
model_lr = Pipeline([
    ("tfidf", word_tfidf),
    ("clf", LogisticRegression(
        max_iter=800,
        n_jobs=-1,
        C=4.0,
        class_weight="balanced"
    )),
])
model_lr.fit(X_all, y_all)
pred_lr = model_lr.predict(X_test)

pd.DataFrame({
    "id": test_df["id"],
    "emotion": pred_lr
}).to_csv("submission_lr_word.csv", index=False)


# ------- Model 2: Char TF-IDF + Linear SVM -------
model_svm = Pipeline([
    ("tfidf", char_tfidf),
    ("clf", LinearSVC(
        C=1.0,
        class_weight="balanced"
    )),
])
model_svm.fit(X_all, y_all)
pred_svm = model_svm.predict(X_test)

pd.DataFrame({
    "id": test_df["id"],
    "emotion": pred_svm
}).to_csv("submission_svm_char.csv", index=False)

print("Saved both submissions.")
```
